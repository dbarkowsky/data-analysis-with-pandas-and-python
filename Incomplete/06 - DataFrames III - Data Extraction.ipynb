# DataFrames III: Data Extraction

```python
import pandas as pd
```

## This Module's Dataset
- This module's dataset is a collection of all James Bond movies.

```python
movies = pd.read_csv("jamesbond.csv")
movies.tail()
```

## The set_index and reset_index Methods
- The index serves as the collection of primary identifiers/labels/entrypoints for the rows.
- The fastest way to extract a row is from a sorted index by position/label.
- Pandas uses index labels/values when merging different objects together.
- The `set_index` method sets an existing column as the index of the `DataFrame`.
- The `reset_index` method uses the standard numeric index as the index of the `DataFrame`.

```python
movies.set_index("Year").head()
```

```python
# NOTE: Setting the index didn't update the original data frame
movies.head()
```

```python
# But this seems to add a weird index series in addition to the regular index
movies.reset_index().head()
```

## Retrieve Rows by Index Position and Index Label
- The `iloc` accessor retrieves one or more rows by index position.
- The `loc` accessor retrieves one or more rows by index label.
- Provide a pair of square brackets after the accessor.
- The accessors accept single values, lists, and slices.

```python
movies.iloc[4]
```

```python
movies.iloc[3:5] # Not inclusive!
```

```python
movies_by_title = movies.set_index("Film")
movies_by_title.head()
```

```python
movies_by_title.loc["Thunderball"]
```

```python
# I expected this to get a range, but it just returns the two. :(
movies_by_title.loc["Goldfinger":"Thunderball"]
```

## Second Arguments to loc and iloc Accessors
- The second value inside the square brackets targets the columns.
- The `loc` accessor accepts index labels for the rows and columns.
- The `iloc` accessor accepts numeric index positions for the rows and columns.

```python
movies_by_title.loc["GoldenEye", ["Year", "Actor"]]
```

```python
movies.iloc[4, 1]
```

```python
movies.iloc[4, [1,3,5]]
```

## Overwrite Value in a DataFrame
- Use the `iloc` or `loc` accessor on the `DataFrame` to target a value, then provide the equal sign and a new value.

```python
movies_by_title.loc["GoldenEye"]
```

```python
movies_by_title.loc["GoldenEye", "Actor"] = "James Dean"
movies_by_title.loc["GoldenEye"]
```

- The `replace` method replaces all occurrences of a value with another value (think of it like "Find and Replace").

```python
# Whoa! I can use dot notation instead of the strings for series. Much nicer.
movies_by_title.Actor = movies_by_title.Actor.replace("James Dean", "Me!")
movies_by_title.loc["GoldenEye"]
```

- To overwrite multiple values in a `DataFrame`, use an accessor on the `DataFrame` itself.
- Pandas uses **Copy-on-Write**: any subset derived from a `DataFrame` behaves as a copy, not a view.
- Because the intermediate result is a copy, the assignment updates the copy — not the original `DataFrame`.
- Pandas raises a `ChainedAssignmentError` to warn you the operation had no effect.
- Accessors like `loc` and `iloc` can accept Boolean Series. Use them to target the values to overwrite.

```python
movies_by_title.loc[movies_by_title.Actor == "Sean Connery", "Actor"] = "Henry Walton Jones Senior"
movies_by_title
```

## Rename Index Labels or Columns in a DataFrame
- The `rename` method accepts a dictionary for either its `columns` or `index` parameters.
- The dictionary keys represent the existing names and the values represent the new names.
- We can replace all columns by overwriting the `DataFrame`'s `columns` attribute.

```python
# This only works because the index is the title.
# Tried with int index and did nothing.
movies_by_title.rename(index={
  "GoldenEye": "Goldeneye"
})
```

```python
movies.rename(columns={
  "Actor":"Who Played Bond"
}).head()
```

```python
# Note it didn't change the original
movies.columns
```

## Delete Rows or Columns from a DataFrame
- The `drop` method deletes one or more rows/columns from a `DataFrame`.
- Pass the `index` or `columns` parameters a list of the column names to remove.

```python
movies_by_title.drop(columns=["Year"]).head()
```

```python
# Just drops the one row
movies_by_title.drop(index=["A View to a Kill"]).head()
```

- The `pop` method removes and returns a single `Series` (it mutates the `DataFrame` in the process).
- Python's `del` keyword also removes a single `Series`.

```python
# This one actually changes the dataframe!
actors = movies.pop("Actor")
actors
```

```python
# No more actor column!
movies.head()
```

## Create Random Sample with the sample Method
- The `sample` method returns a specified one or more random rows from the `DataFrame`.
- Customize the `axis` parameter to extract random columns.

```python
# Single sample
movies_by_title.sample()
```

```python
movies_by_title.sample(3)
```

```python
movies_by_title.sample(frac=0.25) # oh, 25% of the total rows
```

```python
# Any two example columns
movies_by_title.sample(n=2, axis="columns")
```

## The nsmallest and nlargest Methods
- The `nlargest` method returns a specified number of rows with the largest values from a given column.
- The `nsmallest` method returns rows with the smallest values from a given column.
- The `nlargest` and `nsmallest` methods are more efficient than sorting the entire `DataFrame`.

```python
movies_by_title.nlargest(n=4, columns="Box Office")
```

```python
movies_by_title.Budget.nlargest(2) # also works, but returns series, not data frame
```

## The clip Method
- The `clip` method trims values at specified lower and/or upper thresholds.
- Values below `lower` are set to `lower`; values above `upper` are set to `upper`.

```python
# Doesn't do anything for NaN
movies_by_title["Bond Actor Salary"].clip(lower=5, upper=10)
```

## The apply Method with DataFrames
- The `apply` method invokes a function on every column or every row in the `DataFrame`.
- Pass the uninvoked function as the first argument to the `apply` method.

- Pass the `axis` parameter an argument of `"columns"` to invoke the function on every row.
- Pandas will pass in the row's values as a `Series` object. We can use accessors like `loc` and `iloc` to extract the column's values for that row.

```python
movies_by_title = pd.read_csv("jamesbond.csv", index_col="Film").sort_index()
```

```python
def find_sean_connery(movie):
  if movie.loc["Actor"] == "Sean Connery":
    return "Found him!"
  return "He's missing."
```

```python
movies_by_title.apply(find_sean_connery, axis="columns")
```
